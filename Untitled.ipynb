```python
# 统计list中各个item出现的次数
nums = [4,4,4,5,5,6,6]
time_dict = defaultdict(int)
for num in nums:
    time_dict[num] += 1
# 更改字典，key为出现次数，value为相应的数字的集合
index_dict = defaultdict(list)
for key in time_dict:
    index_dict[time_dict[key]].append(key)
# 排序
key = list(index_dict.keys())
```
